# Synthetic readiness demonstration
This is synthetic data only. The target is an unresolved practical-readiness blocker/supportive follow-up need, not a clinical outcome. Feature availability is recorded at T-7, T-2, and T-1. The model is not clinical risk, causal, or calibrated.

A patient-separated split prevents identity leakage. A simple baseline is compared with a fitted logistic model on held-out rows. Limitations are documented in `ml/data/DATA_DICTIONARY.md`.


```python
from pathlib import Path
from ml.train import CHECKPOINTS, MODEL_FEATURES, load_dataset, run_pipeline

output_root = Path(PIPELINE_ROOT)
pipeline_root = output_root / 'ml' if (output_root / 'ml').exists() else output_root
result = run_pipeline(pipeline_root)
rows = load_dataset(result.dataset_path)
evaluation = result.artifact['evaluation']
print('Synthetic rows:', len(rows), 'Feature availability:', ', '.join(MODEL_FEATURES))
print('Patient-separated held-out metrics:', evaluation['model'])
print('Baseline:', evaluation['baseline'])
print('Confusion matrix:', evaluation['model']['confusion_matrix'])
for checkpoint in result.artifact['checkpoints']:
    print(checkpoint['checkpoint'], 'score=', checkpoint['score'], 'factors=', checkpoint['factors'])
comparison = result.artifact['what_if']
print('Transportation sensitivity:', comparison['baseline']['score'], '->', comparison['transportation_available']['score'], 'delta=', comparison['score_delta'])
print('Limitations: synthetic data only; no clinical, causal, or calibrated claims.')
```
